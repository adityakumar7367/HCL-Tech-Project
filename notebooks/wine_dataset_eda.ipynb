## Objective
Perform basic descriptive EDA on the UCI Wine Recognition Dataset. No machine learning is performed.

```python
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

project_root = Path.cwd().resolve()
if project_root.name == "notebooks":
    project_root = project_root.parent
raw_data = project_root / "data" / "raw" / "wine.data"

columns = [
    "Class", "Alcohol", "Malic acid", "Ash", "Alcalinity of ash",
    "Magnesium", "Total phenols", "Flavanoids", "Nonflavanoid phenols",
    "Proanthocyanins", "Color intensity", "Hue",
    "OD280/OD315 of diluted wines", "Proline"
]
```

## Dataset Loading and Description

```python
df = pd.read_csv(raw_data, header=None, names=columns)
df.insert(0, "Record_ID", np.arange(1, len(df) + 1))
print("Loaded source shape:", (len(df), len(columns)))
print("DataFrame shape including Record_ID:", df.shape)
display(df.head())
display(df.tail())
print("Shape:", df.shape)
df.info()
display(df.drop(columns="Record_ID").describe())
```

## 1. Records in Each Wine Class

```python
class_counts = df["Class"].value_counts().sort_index()
display(class_counts.to_frame("Count"))
```

## 2. Alcohol Mean, Median, Minimum, and Maximum

```python
alcohol_array = df["Alcohol"].to_numpy()
alcohol_stats = pd.DataFrame({
    "Statistic": ["Mean", "Median", "Minimum", "Maximum"],
    "Value": [np.mean(alcohol_array), np.median(alcohol_array), np.min(alcohol_array), np.max(alcohol_array)]
})
display(alcohol_stats)
```

## 3. Highest Average Alcohol and Magnesium by Class

```python
alcohol_means = df.groupby("Class")["Alcohol"].mean()
magnesium_means = df.groupby("Class")["Magnesium"].mean()
print("Highest average Alcohol:", int(alcohol_means.idxmax()), alcohol_means.max())
print("Highest average Magnesium:", int(magnesium_means.idxmax()), magnesium_means.max())
```

## 4. Ten Records with the Highest Color Intensity

```python
top10 = df.sort_values("Color intensity", ascending=False).head(10)
display(top10[["Record_ID", "Class", "Color intensity"]])
```

## 5. Missing Values and Duplicate Rows

```python
missing = df.drop(columns="Record_ID").isna().sum()
display(missing.to_frame("Missing Count"))
print("Duplicate rows:", int(df.drop(columns="Record_ID").duplicated().sum()))
```

## 6. Average Alcohol, Malic Acid, and Color Intensity by Class

```python
class_averages = df.groupby("Class")[["Alcohol", "Malic acid", "Color intensity"]].mean()
display(class_averages)
```

## Chart A: Wine Records by Class

```python
plt.figure(figsize=(8, 5))
class_counts.plot(kind="bar", color="#4C78A8", edgecolor="black")
plt.title("Number of Wine Records by Class")
plt.xlabel("Wine Class")
plt.ylabel("Record Count")
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig(project_root / "images" / "01_wine_class_counts.png", dpi=200)
plt.show()
```

## Chart B: Distribution of Alcohol

```python
plt.figure(figsize=(8, 5))
plt.hist(df["Alcohol"], bins=12, color="#F2A65A", edgecolor="black")
plt.title("Distribution of Alcohol Values")
plt.xlabel("Alcohol")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig(project_root / "images" / "02_alcohol_distribution.png", dpi=200)
plt.show()
```
